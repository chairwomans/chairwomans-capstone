```python
# print size 및 validation dataset size 조정

with open('/content/MTA/main.py', 'r') as f:
    code = f.read()

code = code.replace(
    'parser.add_argument(\'-p\', \'--print-freq\', default=200',
    'parser.add_argument(\'-p\', \'--print-freq\', default=64'
)

old = """        val_dataset = build_dataset(set_id, data_transform, args.data, mode=args.dataset_mode)
        print("number of test samples: {}".format(len(val_dataset)))"""

new = """        val_dataset = build_dataset(set_id, data_transform, args.data, mode=args.dataset_mode)
        print("number of test samples: {}".format(len(val_dataset)))
        val_dataset = torch.utils.data.Subset(val_dataset, torch.randperm(len(val_dataset))[:500])
        print("using subset: {} samples".format(len(val_dataset)))"""

code = code.replace(old, new)

with open('/content/MTA/main.py', 'w') as f:
    f.write(code)
```

```python
# batch size 조정 (16)

with open('/content/MTA/main.py', 'r') as f:
    code = f.read()

code = code.replace(
    'parser.add_argument(\'-b\', \'--batch-size\', default=64',
    'parser.add_argument(\'-b\', \'--batch-size\', default=16'
)

with open('/content/MTA/main.py', 'w') as f:
    f.write(code)
```
